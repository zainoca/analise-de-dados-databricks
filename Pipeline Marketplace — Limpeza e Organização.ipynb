## 1. Bronze — dados brutos

Carga "as is" dos CSVs para as tabelas Delta `bronze_vendedores` e `bronze_produtos` (catálogo `workspace`, schema `default`). **Todas as colunas ficam como texto** — nenhuma conversão, correção ou descarte acontece aqui. Cada linha recebe duas colunas de auditoria: `_arquivo_origem` (CSV de origem) e `_carregado_em` (momento da carga).

> A Bronze é a "prova" de tudo que chegou: será o denominador da taxa de aprovação da pergunta 4 (confiabilidade).

```python
from pyspark.sql import functions as F


def carregar_bronze(caminho_csv: str, nome_tabela: str):
    """Carrega um CSV cru para a Bronze: todas as colunas como texto, sem alterar nada."""
    df = (
        spark.read.option("header", True)
        .csv(caminho_csv)
        .withColumn("_arquivo_origem", F.lit(caminho_csv.split("/")[-1]))
        .withColumn("_carregado_em", F.current_timestamp())
    )
    df.write.mode("overwrite").saveAsTable(nome_tabela)
    print(f"{nome_tabela}: {df.count()} linhas carregadas")
    return df


bronze_vendedores = carregar_bronze("/Workspace/projeto1/projeto_residencia/vendedores.csv", "bronze_vendedores")
bronze_produtos = carregar_bronze("/Workspace/projeto1/projeto_residencia/produtos.csv", "bronze_produtos")
```

```python
# Validação: as contagens da Bronze devem bater com os CSVs originais
# (vendedores = 2.067 linhas, produtos = 2.081 linhas — nada pode se perder na carga)
validacao_bronze = spark.sql("""
    SELECT 'vendedores' AS tabela, COUNT(*) AS linhas FROM bronze_vendedores
    UNION ALL
    SELECT 'produtos', COUNT(*) FROM bronze_produtos
""")
display(validacao_bronze)

# Amostra de que a sujeira original foi preservada sem alteração (preços com R$ e vírgula)
display(spark.table("bronze_produtos").where(F.col("preco").startswith("R$")).limit(5))
```

## 2. Silver — limpeza, padronização e quarentena

Regras aplicadas sobre a Bronze:

* **IDs**: `VND-xxxxx` / `PRD-xxxxxx` — `upper` + `trim` + hífen restaurado. Linha sem ID recuperável vai para a quarentena.
* **Duplicatas / versões antigas**: mantida apenas a versão mais recente por ID (`row_number()` ordenando `updated_at desc`); as demais vão para a quarentena com motivo `duplicata_versao_antiga`.
* **Vendedores**: CNPJ só dígitos + flag `cnpj_valido`; e-mail minúsculo + flag `email_valido`; telefone só dígitos sem prefixo `55`; cidade `initcap`; UF normalizada (nome por extenso → sigla); categoria via de-para das **7 categorias canônicas**; data de cadastro em 4 formatos; status → `ativo`/`inativo`/`suspenso`; avaliação com vírgula→ponto, aceita só 0–5.
* **Produtos**: preço e custo limpos de `R$`, ponto de milhar e vírgula decimal; preço nulo/negativo → quarentena; vendedor inexistente ou vazio → quarentena; estoque → inteiro (negativo, `N/A` ou `sem estoque` → nulo); peso → `peso_kg` (gramas ÷ 1000); `ativo` → booleano; data futura → nula + flag `data_futura`; `custo_suspeito` quando custo > preço (a linha fica, mas sai do cálculo de margem na Gold).

Toda linha rejeitada cai em `silver_*_quarentena` com `motivo_rejeicao` — **Silver + quarentena sempre fecha com a Bronze** (base da pergunta 4, confiabilidade).

```python
from functools import reduce
from pyspark.sql import functions as F
from pyspark.sql.window import Window


def limpar_id(colname: str, prefixo: str, nd: int):
    """'vnd-00042' / 'VND00042' / ' VND-00042 ' -> 'VND-00042' (padrão PREFIXO-xxxxx)."""
    padrao = f"^{prefixo}-?(\\d{{{nd}}})$"
    return F.regexp_replace(F.upper(F.trim(F.col(colname))), padrao, f"{prefixo}-$1")


# 7 categorias canônicas: qualquer grafia das 35 variantes cai em uma delas
DE_PARA_CATEGORIA = {
    "beleza": "Beleza", "beleza e saude": "Beleza",
    "casa e cozinha": "Casa e Cozinha", "casa & cozinha": "Casa e Cozinha", "casa/cozinha": "Casa e Cozinha",
    "eletronicos": "Eletrônicos", "eletronico": "Eletrônicos",
    "esporte": "Esporte", "esporte e lazer": "Esporte", "esportes": "Esporte",
    "livros": "Livros", "livro": "Livros",
    "moda": "Moda", "vestuario": "Moda",
    "pet": "Pet", "pets": "Pet", "pet shop": "Pet",
}


def mapear_categoria(colname: str):
    """Normaliza acentos/caixa/espaços e aplica o de-para das categorias canônicas."""
    chave = F.translate(F.lower(F.trim(F.col(colname))), "áàâãéêíóôõúüç", "aaaaeeiooouuc")
    base = F.when(F.length(F.col(colname)) == -1, F.lit(None).cast("string"))  # condição impossível: só inicia o encadeamento
    return reduce(lambda acc, kv: acc.when(chave == F.lit(kv[0]), F.lit(kv[1])), DE_PARA_CATEGORIA.items(), base)


def parse_data_multiformato(colname: str):
    """Converte 'yyyy-MM-dd', 'dd/MM/yyyy', 'yyyy/MM/dd' e 'dd-MM-yyyy' para DATE."""
    d = F.trim(F.col(colname))
    return F.coalesce(
        F.try_to_date(d, "yyyy-MM-dd"), F.try_to_date(d, "dd/MM/yyyy"),
        F.try_to_date(d, "yyyy/MM/dd"), F.try_to_date(d, "dd-MM-yyyy"),
    )


def txt_valor(c):
    """Limpa 'R$ 1.234,56' / '129,31' / '1386.55' -> string numérica com ponto decimal."""
    s = F.regexp_replace(c, "(?i)R\\$|\\s", "")
    return F.when(s.contains(","), F.regexp_replace(F.regexp_replace(s, "\\.", ""), ",", ".")).otherwise(s)
```

```python
# ---- Silver de vendedores: padronização, deduplicação e quarentena ----
br_v = spark.table("bronze_vendedores")
v = br_v.withColumns({
    "seller_id_norm": limpar_id("seller_id", "VND", 5),
    "updated_at_ts": F.try_to_timestamp("updated_at"),
})

# 1) Linhas sem ID recuperável -> quarentena
cond_id_ok = F.coalesce(F.col("seller_id_norm").rlike("^VND-\\d{5}$"), F.lit(False))
sem_id = v.where(~cond_id_ok)
com_id = v.where(cond_id_ok)

# 2) Deduplicação: mantém apenas a versão mais recente de cada vendedor
w_v = Window.partitionBy("seller_id_norm").orderBy(F.col("updated_at_ts").desc())
rank_v = com_id.withColumn("_rn", F.row_number().over(w_v))
versoes_antigas_v = rank_v.where(F.col("_rn") > 1)
atuais_v = rank_v.where(F.col("_rn") == 1)

# 3) Padronização campo a campo
UF_NOME_SIGLA = {"SÃO PAULO": "SP", "DISTRITO FEDERAL": "DF", "BAHIA": "BA",
                 "MINAS GERAIS": "MG", "PARANÁ": "PR", "RIO DE JANEIRO": "RJ"}
uf_up = F.upper(F.trim("uf"))
uf_final = reduce(lambda acc, kv: acc.when(uf_up == F.lit(kv[0]), F.lit(kv[1])),
                  UF_NOME_SIGLA.items(), F.when(F.length(uf_up) == 2, uf_up))

st = F.upper(F.trim("status"))
status_final = (F.when(st.isin("ATIVO", "A"), "ativo")
                .when(st.isin("INATIVO", "I"), "inativo")
                .when(st.isin("SUSPENSO", "S"), "suspenso"))

tel_digitos = F.regexp_replace(F.col("telefone"), "\\D", "")
telefone_final = F.when(tel_digitos.rlike("^55\\d{10,11}$"), F.substring(tel_digitos, 3, 20)).otherwise(tel_digitos)

silver_vendedores = (
    atuais_v
    .withColumns({
        "_cnpj": F.regexp_replace(F.col("cnpj"), "\\D", ""),
        "_email": F.lower(F.trim("email")),
        "_telefone": telefone_final,
        "_uf": uf_final,
        "_categoria": mapear_categoria("categoria_principal"),
        "_data": parse_data_multiformato("data_cadastro"),
        "_status": status_final,
        "_avaliacao_txt": F.regexp_replace(F.trim("avaliacao_media"), ",", "."),
    })
    .withColumn("_avaliacao", F.expr("try_cast(_avaliacao_txt AS DOUBLE)"))
    .select(
        F.col("seller_id_norm").alias("seller_id"),
        F.initcap(F.trim("nome_fantasia")).alias("nome_fantasia"),
        F.trim("razao_social").alias("razao_social"),
        F.col("_cnpj").alias("cnpj"),
        F.coalesce(F.length(F.col("_cnpj")) == 14, F.lit(False)).alias("cnpj_valido"),
        F.col("_email").alias("email"),
        F.coalesce(F.col("_email").rlike("^[^@\\s]+@[^@\\s]+\\.[^@\\s]+$"), F.lit(False)).alias("email_valido"),
        F.col("_telefone").alias("telefone"),
        F.initcap(F.trim("cidade")).alias("cidade"),
        F.col("_uf").alias("uf"),
        F.col("_categoria").alias("categoria_principal"),
        F.col("_data").alias("data_cadastro"),
        F.col("_status").alias("status"),
        F.when((F.col("_avaliacao") >= 0) & (F.col("_avaliacao") <= 5), F.col("_avaliacao")).alias("avaliacao_media"),
        F.col("updated_at_ts").alias("updated_at"),
        F.col("_arquivo_origem"),
    )
)

quarentena_vendedores = (
    sem_id.select(*br_v.columns, F.lit("id_ausente_ou_invalido").alias("motivo_rejeicao"))
    .unionByName(versoes_antigas_v.select(*br_v.columns, F.lit("duplicata_versao_antiga").alias("motivo_rejeicao")))
)

silver_vendedores.write.mode("overwrite").saveAsTable("silver_vendedores")
quarentena_vendedores.write.mode("overwrite").saveAsTable("silver_vendedores_quarentena")
print(f"silver_vendedores: {silver_vendedores.count()} linhas | quarentena: {quarentena_vendedores.count()} linhas")
```

```python
# ---- Silver de produtos: padronização, deduplicação, integridade e quarentena ----
br_p = spark.table("bronze_produtos")
p = br_p.withColumns({
    "product_id_norm": limpar_id("product_id", "PRD", 6),
    "seller_id_norm": limpar_id("seller_id", "VND", 5),
    "updated_at_ts": F.try_to_timestamp("updated_at"),
})

# 1) Linhas sem product_id recuperável
cond_pid_ok = F.coalesce(F.col("product_id_norm").rlike("^PRD-\\d{6}$"), F.lit(False))
sem_pid = p.where(~cond_pid_ok)
com_pid = p.where(cond_pid_ok)

# 2) Deduplicação: mantém apenas a versão mais recente de cada produto
w_p = Window.partitionBy("product_id_norm").orderBy(F.col("updated_at_ts").desc())
rank_p = com_pid.withColumn("_rn", F.row_number().over(w_p))
versoes_antigas_p = rank_p.where(F.col("_rn") > 1)
atuais_p = rank_p.where(F.col("_rn") == 1)

# 3) Integridade relacional: seller_id precisa existir na silver_vendedores
ids_vendedores = spark.table("silver_vendedores").select(F.col("seller_id").alias("seller_id_norm"))
com_vendedor = atuais_p.join(ids_vendedores, "seller_id_norm", "left_semi")
sem_vendedor = atuais_p.join(ids_vendedores, "seller_id_norm", "left_anti")

# 4) Preço precisa ser numérico válido e >= 0
base = com_vendedor.withColumns({
    "_preco_txt": txt_valor(F.col("preco")),
    "_custo_txt": txt_valor(F.col("custo")),
    "_peso_txt": (
        F.when(F.trim("peso").rlike("(?i)\\d+\\s*g$") & ~F.trim("peso").rlike("(?i)kg"),
               F.regexp_replace(F.regexp_replace(F.trim("peso"), "(?i)\\s*g$", ""), ",", "."))
        .when(F.trim("peso").rlike("(?i)kg$"),
              F.regexp_replace(F.regexp_replace(F.trim("peso"), "(?i)\\s*kg$", ""), ",", "."))
        .when(F.trim("peso").rlike("^[0-9.,]+$"), F.regexp_replace(F.trim("peso"), ",", "."))
        .otherwise(F.lit(None).cast("string"))
    ),
    "_eh_grama": F.trim("peso").rlike("(?i)\\d+\\s*g$") & ~F.trim("peso").rlike("(?i)kg"),
    "_estoque_txt": F.trim("estoque"),
    "_data": parse_data_multiformato("data_cadastro"),
})
base = base.withColumns({
    "_preco": F.expr("try_cast(_preco_txt AS DOUBLE)"),
    "_custo": F.expr("try_cast(_custo_txt AS DOUBLE)"),
    "_peso": F.expr("try_cast(_peso_txt AS DOUBLE)"),
    "_estoque": F.expr("try_cast(_estoque_txt AS DOUBLE)"),
    "_data_futura": F.coalesce(F.col("_data") > F.current_date(), F.lit(False)),
})
cond_preco_ok = F.col("_preco").isNotNull() & (F.col("_preco") >= 0)
preco_ruim = base.where(~cond_preco_ok)
aprovados = base.where(cond_preco_ok)

ativo_final = (
    F.when(F.upper(F.trim("ativo")).isin("TRUE", "S", "1", "SIM"), F.lit(True))
    .when(F.upper(F.trim("ativo")).isin("FALSE", "N", "0", "NÃO", "NAO"), F.lit(False))
)

silver_produtos = aprovados.select(
    F.col("product_id_norm").alias("product_id"),
    F.col("seller_id_norm").alias("seller_id"),
    F.trim("sku").alias("sku"),
    F.initcap(F.trim("nome_produto")).alias("nome_produto"),
    mapear_categoria("categoria").alias("categoria"),
    F.coalesce(F.initcap(F.trim("marca")), F.lit("Sem marca")).alias("marca"),
    F.col("_preco").alias("preco"),
    F.col("_custo").alias("custo"),
    F.coalesce(F.col("_custo") > F.col("_preco"), F.lit(False)).alias("custo_suspeito"),
    F.when(F.col("_estoque") < 0, F.lit(None).cast("int")).otherwise(F.col("_estoque").cast("int")).alias("estoque"),
    F.when(F.col("_eh_grama"), F.col("_peso") / 1000).otherwise(F.col("_peso")).alias("peso_kg"),
    ativo_final.alias("ativo"),
    F.when(F.col("_data_futura"), F.lit(None).cast("date")).otherwise(F.col("_data")).alias("data_cadastro"),
    F.col("_data_futura").alias("data_futura"),
    F.col("updated_at_ts").alias("updated_at"),
    F.col("_arquivo_origem"),
)

quarentena_produtos = (
    sem_pid.select(*br_p.columns, F.lit("product_id_ausente_ou_invalido").alias("motivo_rejeicao"))
    .unionByName(versoes_antigas_p.select(*br_p.columns, F.lit("duplicata_versao_antiga").alias("motivo_rejeicao")))
    .unionByName(sem_vendedor.select(*br_p.columns, F.lit("vendedor_ausente_ou_inexistente").alias("motivo_rejeicao")))
    .unionByName(preco_ruim.select(*br_p.columns, F.lit("preco_ausente_ou_invalido").alias("motivo_rejeicao")))
)

silver_produtos.write.mode("overwrite").saveAsTable("silver_produtos")
quarentena_produtos.write.mode("overwrite").saveAsTable("silver_produtos_quarentena")
print(f"silver_produtos: {silver_produtos.count()} linhas | quarentena: {quarentena_produtos.count()} linhas")
```

```python
# ---- Validação da Silver: contagens, motivos de rejeição e taxa de aprovação (pergunta 4) ----
validacao_silver = spark.sql("""
    SELECT tabela, bronze, aprovadas, quarentena,
           ROUND(aprovadas / bronze * 100, 2) AS taxa_aprovacao_pct
    FROM (
        SELECT 'vendedores' AS tabela,
               (SELECT COUNT(*) FROM bronze_vendedores) AS bronze,
               (SELECT COUNT(*) FROM silver_vendedores) AS aprovadas,
               (SELECT COUNT(*) FROM silver_vendedores_quarentena) AS quarentena
        UNION ALL
        SELECT 'produtos',
               (SELECT COUNT(*) FROM bronze_produtos),
               (SELECT COUNT(*) FROM silver_produtos),
               (SELECT COUNT(*) FROM silver_produtos_quarentena)
    )
""")
display(validacao_silver)

display(spark.sql("""
    SELECT 'vendedores' AS tabela, motivo_rejeicao, COUNT(*) AS linhas
    FROM silver_vendedores_quarentena GROUP BY motivo_rejeicao
    UNION ALL
    SELECT 'produtos', motivo_rejeicao, COUNT(*)
    FROM silver_produtos_quarentena GROUP BY motivo_rejeicao
    ORDER BY tabela, linhas DESC
"""))

# Amostra da Silver já padronizada
display(spark.table("silver_produtos").limit(5))
```

## 3. Gold — agregados para as 4 perguntas da diretoria

1. **Concentração**: `gold_q1_concentracao` — ranking de todos os vendedores por quantidade de produtos ativos, com % do catálogo e % acumulado (a resposta é a fatia acumulada dos 10 maiores).
2. **Margem**: `gold_q2_margem_categoria` — margem média `(preço − custo) / preço` por categoria, sem `custo_suspeito` nem custo vazio.
3. **Estoque**: `gold_q3_estoque_uf` — valor em estoque `SUM(preço × estoque)` por UF do vendedor.
4. **Confiabilidade**: `gold_q4_confiabilidade` + `gold_q4_motivos_rejeicao` — taxa de aprovação da Silver sobre a Bronze, por tabela, e o detalhamento por motivo de rejeição.

Cada tabela Gold é um dataset pronto para o dashboard.

```python
# ---- Pergunta 1: quais vendedores concentram o nosso catálogo? ----
# Ranking de todos os vendedores por quantidade de produtos ATIVOS,
# com % do catálogo e % acumulado (resposta = fatia acumulada dos 10 primeiros)
gold_q1_concentracao = spark.sql("""
    WITH catalogo AS (
        SELECT v.seller_id, v.nome_fantasia, v.uf, v.categoria_principal,
               COUNT(p.product_id) AS qtd_produtos_ativos
        FROM silver_vendedores v
        LEFT JOIN silver_produtos p
          ON p.seller_id = v.seller_id AND p.ativo = TRUE
        GROUP BY v.seller_id, v.nome_fantasia, v.uf, v.categoria_principal
    ), total AS (
        SELECT SUM(qtd_produtos_ativos) AS total_ativos FROM catalogo
    )
    SELECT ROW_NUMBER() OVER (ORDER BY qtd_produtos_ativos DESC, seller_id) AS posicao,
           c.seller_id, c.nome_fantasia, c.uf, c.categoria_principal,
           c.qtd_produtos_ativos,
           ROUND(c.qtd_produtos_ativos / t.total_ativos * 100, 2) AS pct_do_catalogo,
           ROUND(SUM(c.qtd_produtos_ativos) OVER (ORDER BY c.qtd_produtos_ativos DESC, c.seller_id)
                 / t.total_ativos * 100, 2) AS pct_acumulado
    FROM catalogo c CROSS JOIN total t
""")
gold_q1_concentracao.write.mode("overwrite").saveAsTable("gold_q1_concentracao")

display(gold_q1_concentracao.where("posicao <= 10").orderBy("posicao"))

fatia_top10 = gold_q1_concentracao.where("posicao = 10").select("pct_acumulado").first()[0]
print(f"Resposta 1: os 10 maiores vendedores concentram {fatia_top10}% dos produtos ativos")
```

```python
# ---- Pergunta 2: qual a margem estimada por categoria? ----
# Margem = (preco - custo) / preco, média por categoria.
# Produtos com custo_suspeito (custo > preço) ou custo vazio ficam fora da conta.
gold_q2_margem = spark.sql("""
    SELECT categoria,
           COUNT(*) AS produtos_analisados,
           ROUND(AVG((preco - custo) / preco) * 100, 2) AS margem_media_pct
    FROM silver_produtos
    WHERE categoria IS NOT NULL
      AND custo IS NOT NULL
      AND NOT custo_suspeito
    GROUP BY categoria
    ORDER BY margem_media_pct DESC
""")
gold_q2_margem.write.mode("overwrite").saveAsTable("gold_q2_margem_categoria")

display(gold_q2_margem)
```

```python
# ---- Pergunta 3: quanto temos em estoque, por estado? ----
# Valor em estoque = SUM(preco * estoque) por UF do vendedor (estoque vazio conta como 0)
gold_q3_estoque = spark.sql("""
    SELECT COALESCE(v.uf, 'Sem UF') AS uf,
           COUNT(DISTINCT v.seller_id) AS vendedores,
           COUNT(p.product_id) AS produtos,
           ROUND(SUM(p.preco * COALESCE(p.estoque, 0)), 2) AS valor_estoque
    FROM silver_produtos p
    JOIN silver_vendedores v ON p.seller_id = v.seller_id
    GROUP BY COALESCE(v.uf, 'Sem UF')
    ORDER BY valor_estoque DESC
""")
gold_q3_estoque.write.mode("overwrite").saveAsTable("gold_q3_estoque_uf")

display(gold_q3_estoque)
```

```python
# ---- Pergunta 4: quanto do nosso cadastro é confiável? ----
# Taxa de aprovação da Silver sobre a Bronze, por tabela, + detalhamento por motivo de rejeição
gold_q4_confiabilidade = spark.sql("""
    SELECT tabela, linhas_bronze, linhas_aprovadas, linhas_quarentena,
           ROUND(linhas_aprovadas / linhas_bronze * 100, 2) AS taxa_aprovacao_pct
    FROM (
        SELECT 'vendedores' AS tabela,
               (SELECT COUNT(*) FROM bronze_vendedores) AS linhas_bronze,
               (SELECT COUNT(*) FROM silver_vendedores) AS linhas_aprovadas,
               (SELECT COUNT(*) FROM silver_vendedores_quarentena) AS linhas_quarentena
        UNION ALL
        SELECT 'produtos',
               (SELECT COUNT(*) FROM bronze_produtos),
               (SELECT COUNT(*) FROM silver_produtos),
               (SELECT COUNT(*) FROM silver_produtos_quarentena)
    )
""")
gold_q4_confiabilidade.write.mode("overwrite").saveAsTable("gold_q4_confiabilidade")

gold_q4_motivos = spark.sql("""
    SELECT 'vendedores' AS tabela, motivo_rejeicao, COUNT(*) AS linhas
    FROM silver_vendedores_quarentena GROUP BY motivo_rejeicao
    UNION ALL
    SELECT 'produtos', motivo_rejeicao, COUNT(*)
    FROM silver_produtos_quarentena GROUP BY motivo_rejeicao
    ORDER BY tabela, linhas DESC
""")
gold_q4_motivos.write.mode("overwrite").saveAsTable("gold_q4_motivos_rejeicao")

display(gold_q4_confiabilidade)
display(gold_q4_motivos)
```
